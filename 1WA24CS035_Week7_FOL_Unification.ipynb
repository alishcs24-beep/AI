```python


def is_variable(x):
    return isinstance(x, str) and x.islower() and x.isalpha()


def occurs_check(var, term, subst):
    term = substitute(term, subst)

    if var == term:
        return True

    if isinstance(term, tuple):
        return any(occurs_check(var, arg, subst) for arg in term)

    return False


def substitute(term, subst):
    if isinstance(term, str):
        while term in subst:
            term = subst[term]
        return term

    if isinstance(term, tuple):
        return tuple(substitute(arg, subst) for arg in term)

    return term


def unify(x, y, subst=None):
    if subst is None:
        subst = {}

    x = substitute(x, subst)
    y = substitute(y, subst)

    if x == y:
        return subst

    if is_variable(x):
        if occurs_check(x, y, subst):
            return None
        subst[x] = y
        return subst

    if is_variable(y):
        if occurs_check(y, x, subst):
            return None
        subst[y] = x
        return subst

    if isinstance(x, tuple) and isinstance(y, tuple):
        if len(x) != len(y) or x[0] != y[0]:
            return None

        for a, b in zip(x[1:], y[1:]):
            subst = unify(a, b, subst)
            if subst is None:
                return None

        return subst

    return None


# Main program
print("FIRST ORDER LOGIC - UNIFICATION")

expr1 = ("P", "x", "a")
expr2 = ("P", "b", "a")

result = unify(expr1, expr2)

if result is not None:
    print("Unification Successful!")
    print("Most General Unifier (MGU):", result)
else:
    print("Unification Failed!")
```
