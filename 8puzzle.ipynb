```python
def dfs(start, goal, limit=30):
    stack = [(start, [start])]
    visited = set()

    while stack:
        state, path = stack.pop()

        if state == goal:
            print("Goal state reached!")
            for step in path:
                print_state(step)
            print("Total moves:", len(path) - 1)
            return True

        if state in visited or len(path) - 1 >= limit:
            continue

        visited.add(state)
        zero = state.index(0)
        row, col = divmod(zero, 3)

        moves = []
        if row > 0:
            moves.append(zero - 3)  # Up
        if row < 2:
            moves.append(zero + 3)  # Down
        if col > 0:
            moves.append(zero - 1)  # Left
        if col < 2:
            moves.append(zero + 1)  # Right

        for pos in moves:
            new_state = list(state)
            new_state[zero], new_state[pos] = (
                new_state[pos], new_state[zero]
            )
            new_state = tuple(new_state)

            if new_state not in visited:
                stack.append((new_state, path + [new_state]))

    print("Solution not found within the depth limit.")
    return False


def print_state(state):
    for i in range(0, 9, 3):
        print(state[i:i + 3])
    print()


start = (1, 2, 3,
         4, 0, 6,
         7, 5, 8)

goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)

print("DFS Solution:")
dfs(start, goal)
```

```python
def depth_limited_search(state, goal, limit, path, visited):
    if state == goal:
        return path

    if len(path) - 1 == limit:
        return None

    visited.add(state)
    zero = state.index(0)
    row, col = divmod(zero, 3)

    moves = []
    if row > 0:
        moves.append(zero - 3)  # Up
    if row < 2:
        moves.append(zero + 3)  # Down
    if col > 0:
        moves.append(zero - 1)  # Left
    if col < 2:
        moves.append(zero + 1)  # Right

    for pos in moves:
        new_state = list(state)
        new_state[zero], new_state[pos] = (
            new_state[pos], new_state[zero]
        )
        new_state = tuple(new_state)

        if new_state not in visited:
            result = depth_limited_search(
                new_state, goal, limit,
                path + [new_state], visited
            )

            if result is not None:
                return result

    return None


def ids(start, goal, max_depth=30):
    for depth in range(max_depth + 1):
        print("Searching at depth:", depth)

        result = depth_limited_search(
            start, goal, depth, [start], set()
        )

        if result is not None:
            print("Goal state reached!")
            for step in result:
                print_state(step)

            print("Total moves:", len(result) - 1)
            return True

    print("Solution not found within the depth limit.")
    return False


def print_state(state):
    for i in range(0, 9, 3):
        print(state[i:i + 3])
    print()


start = (1, 2, 3,
         4, 0, 6,
         7, 5, 8)

goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)

print("IDS Solution:")
ids(start, goal)
```
